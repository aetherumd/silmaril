IMFs describe the probability stars are born with a given mass P(m). It’s very difficult to understand how stars of different masses are formed, but we can estimate the likelihood a star forms with a given mass using our observations of the masses of stars in the Milky Way. 

The most modern models for doing this are the Kroupa broken power law and Chabrier log-normal IMF. (These models are based solely on our observations within the Milky Way and astronomers speculate that there are different IMFs for different galaxies and are influenced by a variety of factors.)

Our goal is identify the IMFs of other galaxies by guessing IMF, generating a sample of stars from it, and them comparing the evolution of luminosities from the sample to the luminosities we get from spectroscopy.

(a) Guess IMF -> Generate sample of stars -> Generate evolution of luminosities (l1)
(b) Spectroscopy Team -> Generate evolution of luminosties (l2)
(c) Compare l1 and l2

### (a.1) Guess IMF

How does the script take an IMF and generate a sample of stars. What parameters from the IMF influence the distribution?
How do we define an IMF?

* m_min - minimum mass for massive stars
* m_max - maximum mass for massive stars
* alpha - power-law index
* mass_frac_in_massive_stars - fraction of mass above 8 M_sum

Other parameters
* N - number of positions (stars) in the simulation each with m_mass

Which of these other parameters do we need to keep constant?
* N should be 1 million for our purposes

```python
class IMF:
    def __init__(self, m_min, m_max, alpha, mass_frac_in_massive_stars, N=1000000):
        self.m_min = m_min
        self.m_max = m_max
        self.alpha = alpha
        self.mass_frac_in_massive_stars = mass_frac_in_massive_stars
        self.N = N

m_min = 0.1
m_max = 100
alpha = -1.3
mass_frac_in_massive_stars = 0.1
N = 1000000
imf = IMF(m_min, m_max, alpha, mass_frac_in_massive_stars, N)
```
